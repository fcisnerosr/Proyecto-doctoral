```python
import pandas as pd

# Load the csv files
resultados_df = pd.read_csv('todos_los_resultados_csv.csv')
design_type_df = pd.read_csv('Elemento_design_type.csv')

print("Archivos cargados exitosamente.")
print("Columnas de todos_los_resultados_csv.csv:", resultados_df.columns.tolist())
print("Columnas de Elemento_design_type.csv:", design_type_df.columns.tolist())
```

```python
# Merge the two dataframes
merged_df = pd.merge(resultados_df, design_type_df, on='Elemento', how='left')

print("DataFrames combinados. Primeras 5 filas del resultado:")
print(merged_df.head())
print("Columnas del DataFrame combinado:", merged_df.columns.tolist())
```

```python
# Save the merged dataframe to the original csv file
merged_df.to_csv('todos_los_resultados_csv.csv', index=False)

print("El archivo 'todos_los_resultados_csv.csv' ha sido actualizado con la nueva columna 'Design_Type'.")
```
