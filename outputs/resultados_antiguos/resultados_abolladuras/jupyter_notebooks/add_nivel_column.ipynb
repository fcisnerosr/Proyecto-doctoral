# Añadir columna 'nivel' a todos_los_resultados_csv.csv

Este notebook añade la columna 'nivel' al archivo `todos_los_resultados_csv.csv` basándose en el mapeo disponible en `elemento_design_type.csv`.

```python
import pandas as pd
import os
```

## 1. Cargar los archivos CSV

```python
# Definir las rutas de los archivos
resultados_csv = 'todos_los_resultados_csv.csv'
elemento_design_csv = 'elemento_design_type.csv'

# Cargar los archivos
df_resultados = pd.read_csv(resultados_csv)
df_elemento_design = pd.read_csv(elemento_design_csv)

print(f"Archivo de resultados cargado: {df_resultados.shape[0]} filas, {df_resultados.shape[1]} columnas")
print(f"Archivo de diseño de elementos cargado: {df_elemento_design.shape[0]} filas, {df_elemento_design.shape[1]} columnas")
```

## 2. Verificar las primeras filas de cada archivo

```python
print("\nPrimeras filas de todos_los_resultados_csv.csv:")
display(df_resultados.head())
```

```python
print("\nPrimeras filas de elemento_design_type.csv:")
display(df_elemento_design.head())
```

## 3. Verificar si la columna 'nivel' ya existe

```python
if 'nivel' in df_resultados.columns:
    print("⚠️  La columna 'nivel' ya existe en el archivo. Será sobrescrita.")
    df_resultados = df_resultados.drop('nivel', axis=1)
else:
    print("✓ La columna 'nivel' no existe. Se procederá a añadirla.")
```

## 4. Crear el diccionario de mapeo Elemento -> nivel

```python
# Verificar y limpiar los nombres de las columnas
print("Columnas en df_elemento_design antes de limpiar:")
print(df_elemento_design.columns.tolist())
print(f"\nNombres de columnas (representación): {repr(df_elemento_design.columns.tolist())}")

# Limpiar espacios en blanco de los nombres de columnas
df_elemento_design.columns = df_elemento_design.columns.str.strip()
df_resultados.columns = df_resultados.columns.str.strip()

print("\nColumnas en df_elemento_design después de limpiar:")
print(df_elemento_design.columns.tolist())
```

```python
# Crear diccionario de mapeo: Elemento -> nivel
# Nota: La columna en el CSV puede llamarse 'nivel' o 'Story'
nivel_col = 'nivel' if 'nivel' in df_elemento_design.columns else 'Story'
print(f"Usando la columna '{nivel_col}' del archivo elemento_design_type.csv")

elemento_nivel_map = dict(zip(df_elemento_design['Elemento'], df_elemento_design[nivel_col]))

print(f"\nDiccionario de mapeo creado con {len(elemento_nivel_map)} elementos")
print("\nEjemplos del mapeo:")
for i, (elem, nivel) in enumerate(list(elemento_nivel_map.items())[:10]):
    print(f"  Elemento {elem} -> nivel '{nivel}'")
```

## 5. Añadir la columna 'nivel' al DataFrame de resultados

```python
# Mapear la columna 'Elemento' a 'nivel'
df_resultados['nivel'] = df_resultados['Elemento'].map(elemento_nivel_map)

# Verificar si hay valores nulos (elementos sin nivel asignado)
elementos_sin_nivel = df_resultados[df_resultados['nivel'].isna()]['Elemento'].unique()

if len(elementos_sin_nivel) > 0:
    print(f"⚠️  Advertencia: {len(elementos_sin_nivel)} elementos no tienen nivel asignado:")
    print(elementos_sin_nivel)
else:
    print("✓ Todos los elementos tienen un nivel asignado correctamente.")

print(f"\nColumna 'nivel' añadida exitosamente. Total de columnas: {df_resultados.shape[1]}")
```

## 6. Verificar el resultado

```python
print("\nPrimeras filas del DataFrame actualizado:")
display(df_resultados.head(10))
```

```python
# Mostrar la distribución de niveles
print("\nDistribución de registros por nivel:")
print(df_resultados['nivel'].value_counts().sort_index())
```

## 7. Guardar el archivo actualizado

```python
# Crear backup del archivo original
backup_filename = 'todos_los_resultados_csv_backup.csv'
if not os.path.exists(backup_filename):
    df_original = pd.read_csv(resultados_csv)
    df_original.to_csv(backup_filename, index=False)
    print(f"✓ Backup creado: {backup_filename}")
else:
    print(f"ℹ️  El backup ya existe: {backup_filename}")
```

```python
# Guardar el archivo actualizado
df_resultados.to_csv(resultados_csv, index=False)
print(f"\n✓ Archivo actualizado guardado exitosamente: {resultados_csv}")
print(f"  Total de filas: {df_resultados.shape[0]}")
print(f"  Total de columnas: {df_resultados.shape[1]}")
print(f"\nColumnas del archivo:")
print(list(df_resultados.columns))
```

## 8. Verificación final

```python
# Recargar el archivo para verificar que se guardó correctamente
df_verificacion = pd.read_csv(resultados_csv)

print("\nVerificación del archivo guardado:")
print(f"Columnas: {list(df_verificacion.columns)}")
print(f"\nÚltimas 5 filas del archivo actualizado:")
display(df_verificacion.tail())

if 'nivel' in df_verificacion.columns:
    print("\n✅ ¡Proceso completado exitosamente! La columna 'nivel' ha sido añadida.")
else:
    print("\n❌ Error: La columna 'nivel' no se encuentra en el archivo guardado.")
```
